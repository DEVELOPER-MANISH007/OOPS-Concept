# Polymorphism

**Polymorphism** means "many forms".

In OOP, polymorphism allows the same method, function, or interface to behave differently for different objects.

### Example

```python
class Dog:
    def sound(self):
        print("Bark")


class Cat:
    def sound(self):
        print("Meow")

````python
class Dog:

    def sound(self):
        print("Dog says: Bark")


class Cat:

    def sound(self):
        print("Cat says: Meow")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
````

## Duck Typing

Duck typing is a Python approach where the behavior of an object is more important than its exact type.

The idea is:

> If an object provides the required method or behavior, it can be used.

Example:

```python

    def sound(self):
        print("Bark")

## Operator Overloading

**Operator overloading** allows us to define how operators behave with objects of our custom classes.

Python uses special/dunder methods for operators.

### Common Operators

| Operator | Special Method |
|---|---|
| `+` | `__add__()` |
| `-` | `__sub__()` |
| `*` | `__mul__()` |
| `/` | `__truediv__()` |
| `==` | `__eq__()` |
| `<` | `__lt__()` |
| `>` | `__gt__()` |

### Example

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

````python
class Number:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

n1 = Number(10)
n2 = Number(20)

print(n1 + n2)
````

````python
class Data:
    def __init__(self, value):
        self.value = value

    def __sub__(self, other):
        return self.value - other.value

d1 = Data(100)
d2 = Data(40)

print(d1 - d2)
````

````python
class Data:
    def __init__(self, value):
        self.value = value

    def __mul__(self, other):
        return self.value * other.value

d1 = Data(100)
d2 = Data(40)

print(d1 * d2)
````

## Method Overloading

Method overloading means having multiple methods with the same name but different parameters.

In languages like Java/C++, multiple overloaded methods can exist.

Python does **not** support traditional method overloading directly.

If multiple methods with the same name are defined in a class, the latest definition replaces the previous one.

### Python alternatives

Python commonly achieves overloading-like behavior using:

- Default arguments
- `*args`
- `**kwargs`

### Example using Default Arguments

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

### 1. Default Arguments

````python
class Calculator:

    def add(self, a, b, c=0):
        return a + b + c

calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
````

### 2. *args

````python
class Calculator:

    def add(self, *args):
        return sum(args)


calc = Calculator()

calc.add(10, 20)
calc.add(10, 20, 30)
calc.add(10, 20, 30, 40)
````

## Polymorphism with Inheritance

Polymorphism can be achieved through method overriding in inheritance.

A parent class defines a common method, and child classes provide their own implementations.

Example:

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog: Bark")


class Cat(Animal):
    def sound(self):
        print("Cat: Meow")

## 🏆 Final Challenge — ML Model System

````python
class Model:
    def predict(self):
        return print("Generic model prediction")
class LinearRegression(Model):
    def predict(self):
       return print("Linear Regression prediction")
class DecisionTree(Model):
    def predict(self):
        return print("Decision Tree prediction")
class RandomForest(Model):
    def predict(self):
        return print("Random Forest prediction")
class KNN(Model):
    def predict(self):
        return print("KNN prediction")

def run_model(model):
    model.predict()

lr = LinearRegression()
dt = DecisionTree()
rf = RandomForest()
kn = KNN()

run_model(lr)
run_model(dt)
run_model(rf)
run_model(kn)
````
