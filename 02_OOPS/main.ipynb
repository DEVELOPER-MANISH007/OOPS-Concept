# Inheritance

**Inheritance** is an OOP mechanism in which a child class can reuse and extend the attributes and methods of a parent class.

### Terminology

- **Parent Class** → Class whose features are inherited.
- **Base Class** → Another name for parent class.
- **Super Class** → Another name for parent class.
- **Child Class** → Class that inherits from another class.
- **Derived Class** → Another name for child class.
- **Sub Class** → Another name for child class.

## super()

`super()` is used to access methods and functionality of a parent class from a child class.

It is commonly used when the child class defines its own constructor but also wants to initialize the parent class attributes.

## 2. Multilevel Inheritance

Multilevel inheritance occurs when a class inherits from another derived class, creating an inheritance chain.

### Structure

```text
Grandparent
    ↓
  Parent
    ↓
  Child

````python
class Person:
    pass

class Employee(Person):
    pass

class DataScientist(Employee):
    pass
````

````python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_person(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")


class Employee(Person):

    def __init__(self, name, age, salary):
        super().__init__(name, age)
        self.salary = salary

    def show_employee(self):
        print(f"Salary: {self.salary}")


class DataScientist(Employee):

    def __init__(self, name, age, salary, domain):
        super().__init__(name, age, salary)
        self.domain = domain

    def show_data_scientist(self):
        print(f"Domain: {self.domain}")

ds = DataScientist(
    "Manish",
    21,
    50000,
    "Machine Learning"
)
ds.show_data_scientist()
ds.show_person()
ds.show_employee()
````

## 3. Multiple Inheritance

**Multiple inheritance** occurs when one child class inherits from more than one parent class.

### Structure

```text
Parent 1       Parent 2
    \             /
     \           /
        Child

````python
class Father:

    def driving(self):
        print("Father: Driving")


class Mother:

    def cooking(self):
        print("Mother: Cooking")


class Child(Father, Mother):
    pass

child = Child()

child.driving()
child.cooking()
````

## 4. Hierarchical Inheritance

**Hierarchical inheritance** occurs when multiple child classes inherit from the same parent class.

### Structure

```text
       Parent
       /    \
      ↓      ↓
   Child1  Child2

````python
class Vehicle:

    def start(self):
        print("Vehicle started")


class Car(Vehicle):
    def drive(self):
        print("Car is driving")


class Bike(Vehicle):
    def ride(self):
        print("Bike is riding")

car = Car()
bike = Bike()

car.start()
car.drive()

bike.start()
bike.ride()
````

## 5. Hybrid Inheritance

Hybrid inheritance occurs when multiple types of inheritance are combined in a single inheritance structure.

### Example

```text
          Person
         /      \
    Employee   Skills
         \      /
       DataScientist

````python
class Person:

    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(f"Hello, I am {self.name}")


class Employee(Person):

    def work(self):
        print("Employee is working")


class Skills(Person):

    def learn(self):
        print("Learning new skills")


class DataScientist(Employee, Skills):
    pass

ds = DataScientist("Manish")
ds.introduce()
print(DataScientist.mro())
````
