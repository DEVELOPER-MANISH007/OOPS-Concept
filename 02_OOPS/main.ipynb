# Inheritance

**Inheritance** is an OOP mechanism in which a child class can reuse and extend the attributes and methods of a parent class.

### Terminology

- **Parent Class** → Class whose features are inherited.
- **Base Class** → Another name for parent class.
- **Super Class** → Another name for parent class.
- **Child Class** → Class that inherits from another class.
- **Derived Class** → Another name for child class.
- **Sub Class** → Another name for child class.

## super()

`super()` is used to access methods and functionality of a parent class from a child class.

It is commonly used when the child class defines its own constructor but also wants to initialize the parent class attributes.

### Example

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed


class Car(Vehicle):
    def __init__(self, brand, speed, fuel_type):
        super().__init__(brand, speed)
        self.fuel_type = fuel_type

````python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed


class Car(Vehicle):
    def __init__(self, brand, speed, fuel_type):
        super().__init__(brand, speed)
        self.fuel_type = fuel_type


car1 = Car("Audi", 334, "Diesel")
print(car1.brand)
print(car1.speed)
print(car1.fuel_type)
````

## Method Overriding

**Method overriding** occurs when a child class provides its own implementation of a method that already exists in the parent class.

### Example

```python
class Vehicle:
    def start(self):
        print("Vehicle is starting")


class Car(Vehicle):
    def start(self):
        print("Car is starting with a key")

````python
class Vehicle:
    def start(self):
        print("Vehicle is starting")


class Car(Vehicle):
    def start(self):
        print("Car is starting with a key")

vehicle = Vehicle()
car = Car()

vehicle.start()
car.start()
````
