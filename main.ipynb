# 08 - OOP: Operator Overloading

Operator overloading allows us to define how operators like `+`, `-`, `==`, `<`, and `>` behave for our custom objects.

Python implements operator overloading using special (dunder) methods.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __add__(self, other):
        return self.marks + other.marks

    def __eq__(self, other):
        return self.marks == other.marks

    def __lt__(self, other):
        return self.marks < other.marks

    def __str__(self):
        return f"{self.name} - {self.marks} marks"
```

```python
student1 = Student("Manish", 85)
student2 = Student("Rahul", 75)

print(student1)
print(student2)
```

```python
total_marks = student1 + student2

print("Total Marks:", total_marks)
```

```python
print("Are marks equal?", student1 == student2)
print("Is Student 1 lower than Student 2?", student1 < student2)
```

## Practical Data Science Example

```python
class DataScienceProject:
    def __init__(self, name, accuracy):
        self.name = name
        self.accuracy = accuracy

    def __gt__(self, other):
        return self.accuracy > other.accuracy

    def __str__(self):
        return f"{self.name}: {self.accuracy}% accuracy"
```

```python
project1 = DataScienceProject("Fetal Health Classifier", 94.2)
project2 = DataScienceProject("Sentiment Analysis", 91.8)

print(project1)
print(project2)

if project1 > project2:
    print(f"{project1.name} has better accuracy.")
else:
    print(f"{project2.name} has better accuracy.")
```

## Key Takeaways

- Operator overloading allows operators to work with custom objects.
- Python uses dunder methods for operator overloading.
- `__add__()` → `+`
- `__eq__()` → `==`
- `__lt__()` → `<`
- `__gt__()` → `>`
- `__str__()` → `print(object)`

Operator overloading makes custom classes more intuitive and Pythonic.
