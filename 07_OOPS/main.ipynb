# Composition

Composition represents a strong HAS-A relationship between classes.

One class contains an object of another class.

Ek class ke object ka doosri class ke object se relation hai, lekin dono objects independently exist kar sakte hain.
Example:

```python
class Engine:

    def start(self):
        print("Engine Started")


class Car:

    def __init__(self):
        self.engine = Engine()

````python
class Engine:

    def start(self):
        print("Engine Started")


class Car:

    def __init__(self):
        self.engine = Engine()                      # object user kiya hai dusri class ka yaho composition hai

    def start_car(self):
        self.engine.start()
        print("Car Started")

car = Car()
car.start_car()
````

# Aggregation

Aggregation is a **HAS-A relationship** where one class uses an object of another class, but both objects can exist independently.

Ek class ke paas doosri class ka object hota hai, lekin woh object us class ke andar create nahi hota aur dono independently exist kar sakte hain.
### Example

```python
class Model:

    def train(self):
        print("Model is training")


class DataScientist:

    def __init__(self, model):
        self.model = model

    def train_model(self):
        self.model.train()

````python
class Teacher:
    def __init__(self, name):
        self.name = name

    def teach(self):
        print(f"{self.name} is teaching")


class College:
    def __init__(self, teacher):
        self.teacher = teacher

    def conduct_class(self):
        self.teacher.teach()


# Teacher independently created
teacher1 = Teacher("Manish")

# Existing Teacher object passed to College
college1 = College(teacher1)

college1.conduct_class()
````

| Feature               | Composition       | Aggregation   |
| --------------------- | ----------------- | ------------- |
| Relationship          | HAS-A             | HAS-A         |
| Object created by     | Container class   | Outside class |
| Ownership             | Strong            | Weak          |
| Independent existence | Usually dependent | Independent   |
| Example               | Car → Engine      | Team → Player |



**Bas ek line mein yaad rakh:**  
👉 **Composition = "main banaunga"**  
👉 **Aggregation = "tu bana ke de, main use karunga"** 😎🔥

## 3️⃣ Composition + Polymorphism

````python
class Model:
    def train(self):
        print("Generic Model is Training")

class LinearRegression(Model):
    def train(self):
        print("Training Linear Regression")

class RandomForest(Model):
    
    def train(self):
        print("Training Random Forest")

class DataScientist:
    
    def __init__(self,model):
        self.model = model
    def train_model(self):
        self.model.train()
        
        

lr = LinearRegression()
rf = RandomForest()

ds1 = DataScientist(lr)
ds2 = DataScientist(rf)

ds1.train_model()
ds2.train_model()



````

````python
class Model:
    def train(self):
        print("Generic Model is Training")
    
    def predict(self):
        print("Generic Model is Predicting")

class LinearRegression(Model):
    def train(self):
            print("Linear Regression Model is Training")
        
    def predict(self):
            print("Linear Regression is Predicting")


class RandomForest(Model):
    def train(self):
                print("RandomForest Model is Training")
            
    def predict(self):
                print("RandomForest is Predicting")  

class DataScientist:
    
    def __init__(self,model):
        self.model = model
        
    def train(self):
            self.model.train()
        
    def predict(self):
           self.model.predict()


lr = LinearRegression()
rf = RandomForest()

ds1 = DataScientist(lr)
ds2 = DataScientist(rf)

ds1.train()
ds1.predict()

ds2.train()
ds2.predict()
````

## 4️⃣ Composition — Object internally create hoga

````python
class Department:
    def show_department(self):
        print("Data Science Department")

class Company:
    def __init__(self):
        self.department = Department()

    def show_company_department(self):
        self.department.show_department()

company = Company()

company.show_company_department()
````
