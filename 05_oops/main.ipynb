# Encapsulation

Encapsulation is the process of bundling data and methods together inside a class and controlling access to the internal data.

```
Bank Account
│
├── balance       🔒
├── deposit()
├── withdraw()
└── get_balance()

```
### Example

```python
class Bank:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, balance):
        if balance >= 0:
            self.__balance = balance
        else:
            print("Invalid balance")

````python
class Bank:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, balance):
        if balance >= 0:
            self.__balance = balance
        else:
            print("Invalid balance")

account = Bank(5000)

print(account.get_balance())

account.set_balance(7000)

print(account.get_balance())
````

````python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Invalid withdrawal")

    def get_balance(self):
        return self.__balance


account = BankAccount(5000)

account.deposit(2000)

account.withdraw(1000)

print(account.get_balance())
````

## Read-Only Property

A read-only property can be accessed but cannot be modified directly.

It is created using `@property` without defining a setter.

Example:

```python
class Employee:

    def __init__(self, employee_id):
        self.__employee_id = employee_id

    @property
    def employee_id(self):
        return self.__employee_id

````python
class Employee:

    def __init__(self, employee_id):
        self.__employee_id = employee_id

    @property
    def employee_id(self):
        return self.__employee_id

emp = Employee(101)
print(emp.employee_id)

````

## Encapsulation with Inheritance

A parent class can encapsulate private data and expose it through properties.

Child classes can inherit these properties without directly accessing the private attribute.

Example:

```python
class Person:

    def __init__(self, age):
        self.__age = age

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, value):
        if value >= 0:
            self.__age = value
        else:
            print("Invalid Age")

````python
class Person:

    def __init__(self, name):
        self.__name = name

    @property
    def name(self):
        return self.__name

class DataScientist(Person):
    pass

ds = DataScientist("Manish")

print(ds.name)
````

````python
class Person:

    def __init__(self, age):
        self.__age = age

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, value):
        if value >= 0:
            self.__age = value
        else:
            print("Invalid Age")

class Student(Person):
    pass

student = Student(21)

print(student.age)

student.age = 25

print(student.age)

student.age = -5
````

````python
class BankAccount:
    def __init__(self,balance,pin):
        self.__balance = balance
        self.__pin = pin
    
    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self,value):
        if value >= 0:
            self.__balance = value
        else:
            print("Invalid Balance")
    
    @property
    def pin(self):
        return self.__pin
    
    def withdraw(self,amount):
        if self.__balance >= amount and amount>0:
            self.__balance -= amount
            return self.__balance
        else:
            print("Insufficient balance")
                
    def deposit(self,amount):
        if amount>0:
            self.__balance = self.__balance+amount
            return self.__balance
        else:
            print("Enter Valid Amount")

account = BankAccount(5000, 1234)

print(account.balance)
print(account.pin)

account.deposit(2000)
print(account.balance)

account.withdraw(3000)
print(account.balance)

account.withdraw(10000)

account.deposit(-500)

account.balance = -100

print(account.balance)
````
