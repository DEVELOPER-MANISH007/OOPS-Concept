## 1. What is OOP?

**Object-Oriented Programming** is a programming approach where programs are designed using **objects** and **classes**.

- **Class:** A blueprint/template for creating objects.
- **Object:** An instance of a class.
- **Attribute:** Data stored inside an object.
- **Method:** A function defined inside a class.


``````python
class Student:
    pass

# Creating an object
student1 = Student()

print(type(student1))
``````

## 2. Constructor and Instance Variables

The `__init__()` method is called automatically when an object is created. It is commonly used to initialize object attributes.

``````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name} and I am {self.age} years old.")


student1 = Student("Rahul", 20)
student2 = Student("Priya", 21)

student1.introduce()
student2.introduce()
``````

## 3. Instance Methods

An instance method works with the data of a particular object. The first parameter is conventionally named `self`.

``````python
class Calculator:
    def add(self, a, b):
        return a + b

    def multiply(self, a, b):
        return a * b


calc = Calculator()
print("Addition:", calc.add(10, 5))
print("Multiplication:", calc.multiply(10, 5))
``````

## 4. Encapsulation

Encapsulation means keeping data and the methods that operate on that data together, while controlling how the data is accessed.

Python commonly uses `_name` for protected-by-convention attributes and `__name` for name-mangled private attributes.

``````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance


account = BankAccount("Amit", 5000)
account.deposit(1500)
print("Balance:", account.get_balance())
``````

## 5. Inheritance

Inheritance allows one class to reuse and extend the functionality of another class.

**Parent/Base class → Child/Derived class**

``````python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


dog = Dog()
dog.eat()   # inherited method
dog.bark()  # child-specific method
``````

## 6. Method Overriding

A child class can provide its own implementation of a method already defined in the parent class.

``````python
class Animal:
    def sound(self):
        print("Some animal sound")


class Dog(Animal):
    def sound(self):
        print("Woof!")


class Cat(Animal):
    def sound(self):
        print("Meow!")


Dog().sound()
Cat().sound()
``````

## 7. Polymorphism

**Polymorphism** means that the same interface or method name can behave differently for different objects.

``````python
animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()
``````

## 8. Abstraction

Abstraction means exposing the important interface while hiding implementation details.

Python provides the `abc` module for creating abstract base classes.

``````python
from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


rectangle = Rectangle(10, 5)
print("Area:", rectangle.area())
``````

## 9. Class Variables and Class Methods

A **class variable** is shared by all objects of the class. A `@classmethod` receives the class as its first argument, conventionally called `cls`.

``````python
class Employee:
    company = "ABC Technologies"
    employee_count = 0

    def __init__(self, name):
        self.name = name
        Employee.employee_count += 1

    @classmethod
    def total_employees(cls):
        return cls.employee_count


e1 = Employee("Ravi")
e2 = Employee("Neha")

print(Employee.company)
print("Total employees:", Employee.total_employees())
``````

## 10. The Four Main OOP Principles

| Principle | Meaning |
|---|---|
| **Encapsulation** | Bundle data and methods and control access to data |
| **Inheritance** | Create a new class from an existing class |
| **Polymorphism** | Same interface/method can have different behavior |
| **Abstraction** | Hide implementation details and expose essential behavior |

## 11. Mini Project: Simple Library System

This example combines classes, objects, encapsulation, methods, and object state.

``````python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_available = True

    def borrow(self):
        if self.is_available:
            self.is_available = False
            print(f"'{self.title}' borrowed successfully.")
        else:
            print(f"'{self.title}' is not available.")

    def return_book(self):
        self.is_available = True
        print(f"'{self.title}' returned successfully.")


book = Book("Python Basics", "John Smith")
book.borrow()
book.borrow()
book.return_book()
book.borrow()
``````
