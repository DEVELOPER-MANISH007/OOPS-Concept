## @classmethod

A class method receives the class as its first argument using `cls`.

Syntax:

```python

def method(cls, ...):
    ...

### 1️⃣ Class Method vs Static Method — Practical Use

```python 
@classmethod

````python
class Employee:

    company = "Google"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

print(Employee.company)

Employee.change_company("Microsoft")

print(Employee.company)
````

````python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @classmethod
    def from_string(cls, data):
        name, salary = data.split(",")
        return cls(name, int(salary))

emp = Employee.from_string("Manish,50000")

print(emp.name)
print(emp.salary)
````

# @staticmethod

A static method is a method that does not require access to the instance (`self`) or class (`cls`).

It is defined using `@staticmethod`.

Example:

```python
class Student:

    @staticmethod
    def is_valid_marks(marks):
        return 0 <= marks <= 100

````python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    @staticmethod
    def is_valid_marks(marks):
        return 0 <= marks <= 100
student = Student("Manish", 85)

print(student.is_valid_marks(85))
print(student.is_valid_marks(150))
````

| Method          | First Parameter | Kis cheez se kaam |
| --------------- | --------------- | ----------------- |
| Instance Method | `self`          | Object            |
| Class Method    | `cls`           | Class             |
| Static Method   | None            | Independent logic |

## ``4️⃣ Custom Exception``

# Custom Exceptions

Custom exceptions allow us to create application-specific errors.

A custom exception is created by inheriting from `Exception`.

Example:

```python
class InsufficientBalanceError(Exception):
    pass

````python
class InsufficientBalanceError(Exception):
    pass


class BankAccount:

    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):

        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient balance")

        self.balance -= amount
        print(f"Remaining Balance: {self.balance}")


account = BankAccount(5000)

account.withdraw(2000)
````

# Interface-Style Design in Python

Python does not have a separate `interface` keyword like Java.
We can create interface-like behavior using an **Abstract Base Class (ABC)**.

## Abstract Interface

An abstract class defines the methods that child classes **must implement**.

Example:

```python
from abc import ABC, abstractmethod

class DataProcessor(ABC):

    @abstractmethod
    def clean_data(self):
        pass

    @abstractmethod
    def transform_data(self):
        pass

````python
from abc import ABC, abstractmethod

class Model(ABC):

    @abstractmethod
    def train(self):
        pass

    @abstractmethod
    def predict(self):
        pass

class LinearRegression(Model):

    def train(self):
        print("Training Linear Regression")

    def predict(self):
        print("Predicting using Linear Regression")
class RandomForest(Model):

    def train(self):
        print("Training Random Forest")

    def predict(self):
        print("Predicting using Random Forest")


lr = LinearRegression()
rf = RandomForest()

lr.train()
lr.predict()

rf.train()
rf.predict()
````

````python
from abc import ABC, abstractmethod

class DataProcessor(ABC):
    
    @abstractmethod
    def clean_data(self):
        pass
    
    @abstractmethod
    def transform_data(self):
        pass

class PandasProcessor(DataProcessor):
    def clean_data(self):
            pass
    def transform_data(self):
            pass


class NumpyProcessor(DataProcessor):
    def clean_data(self):
            pass
    def transform_data(self):
            pass


p = PandasProcessor()
n = NumpyProcessor()

p.clean_data()
p.transform_data()

n.clean_data()
n.transform_data()
````
