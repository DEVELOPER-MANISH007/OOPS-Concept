# 09 - OOP: Multiple Inheritance & MRO

Multiple inheritance allows a class to inherit features from more than one parent class.

**MRO (Method Resolution Order)** defines the order in which Python searches for methods and attributes.

## 1. Parent Classes

We create separate classes for data preprocessing and model training.

```python
class DataPreprocessor:
    def preprocess(self):
        print("Dataset preprocessing completed.")


class ModelTrainer:
    def train(self):
        print("Machine Learning model training completed.")
```

## 2. Multiple Inheritance

`MLPipeline` inherits from both `DataPreprocessor` and `ModelTrainer`.

```python
class MLPipeline(DataPreprocessor, ModelTrainer):
    def run(self):
        self.preprocess()
        self.train()
        print("ML pipeline executed successfully.")
```

```python
pipeline = MLPipeline()

pipeline.run()
```

## 3. Understanding MRO

Python provides the `mro()` method and `__mro__` attribute to inspect the Method Resolution Order.

```python
print(MLPipeline.mro())
print("\nMRO using __mro__:")
print(MLPipeline.__mro__)
```

## 4. Method Conflict Example

If multiple parent classes contain a method with the same name, Python follows the MRO to decide which method should run.

```python
class Classification:
    def predict(self):
        print("Classification prediction")


class Regression:
    def predict(self):
        print("Regression prediction")


class MLModel(Classification, Regression):
    pass


model = MLModel()
model.predict()
```

```python
print(MLModel.mro())
```

## Key Takeaways

- Multiple inheritance allows one class to inherit from multiple parent classes.
- Python uses MRO to determine the method lookup order.
- `ClassName.mro()` returns the Method Resolution Order.
- `__mro__` provides the same information as a tuple.
- When methods have the same name in multiple parents, MRO decides which implementation is used.

### Real-world idea

An ML pipeline can combine independent responsibilities such as preprocessing, feature engineering, model training, and evaluation into reusable classes.
