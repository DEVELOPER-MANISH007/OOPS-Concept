# 08 - OOP: Composition & Aggregation

Composition and Aggregation are relationships between classes that help us build complex applications using reusable objects.

- **Composition:** A class strongly owns another object.
- **Aggregation:** A class uses objects that can exist independently.

## 1. Dataset Class

```python
class Dataset:
    def __init__(self, name, rows, columns):
        self.name = name
        self.rows = rows
        self.columns = columns

    def info(self):
        return f"{self.name}: {self.rows} rows, {self.columns} columns"
```

## 2. Model Class

```python
class Model:
    def __init__(self, name, algorithm):
        self.name = name
        self.algorithm = algorithm

    def train(self):
        print(f"{self.name} ({self.algorithm}) is training...")
```

## 3. Composition

`MLProject` creates its own `Dataset` object. This represents **Composition** because the dataset is strongly associated with the project.

```python
class MLProject:
    def __init__(self, project_name, dataset_name):
        self.project_name = project_name

        # Composition
        self.dataset = Dataset(
            dataset_name,
            rows=5000,
            columns=12
        )

    def run(self):
        print(f"Project: {self.project_name}")
        print(self.dataset.info())
```

```python
project = MLProject(
    "Customer Churn Prediction",
    "Customer Dataset"
)

project.run()
```

## 4. Aggregation

`MLTeam` receives existing `Model` objects. The models can exist independently of the team, so this represents **Aggregation**.

```python
class MLTeam:
    def __init__(self, team_name, models):
        self.team_name = team_name
        self.models = models

    def train_all(self):
        print(f"Team: {self.team_name}")

        for model in self.models:
            model.train()
```

```python
model1 = Model(
    "Random Forest",
    "Random Forest Classifier"
)

model2 = Model(
    "Logistic Regression",
    "Logistic Regression"
)

team = MLTeam(
    "Data Science Team",
    [model1, model2]
)

team.train_all()
```

## 5. Composition vs Aggregation

| Concept | Meaning | Example |
|---|---|---|
| Composition | Strong ownership relationship | MLProject → Dataset |
| Aggregation | Weak / independent relationship | MLTeam → Model |

### Key Takeaways

- Composition represents a strong **has-a** relationship.
- Aggregation represents a more independent **has-a** relationship.
- Both concepts help organize large applications into reusable classes.
- These relationships are commonly used in real-world software and ML systems.
