# Magic / Dunder Methods

Magic methods are special methods in Python whose names start and end with double underscores.

Examples:

- `__init__()`
- `__str__()`
- `__repr__()`
- `__len__()`
- `__eq__()`
- `__add__()`

## 2️⃣ __repr__() — Developer Representation

## __repr__()

`__repr__()` provides a developer-oriented representation of an object.

It is generally intended to be more precise and useful for debugging.

Example:

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def __str__(self):
        return f"Employee: {self.name}, Salary: {self.salary}"

    def __repr__(self):
        return f"Employee('{self.name}', {self.salary})"

````python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def __str__(self):
        return f"Employee: {self.name}, Salary: {self.salary}"

    def __repr__(self):
        return f"Employee('{self.name}', {self.salary})"

emp = Employee("Manish", 50000)

print(emp)
print(repr(emp))
````

## __len__()

`__len__()` defines what `len(object)` should return for a custom object.

Example:

```python
class Employee:

    def __init__(self, name, skills):
        self.name = name
        self.skills = skills

    def __len__(self):
        return len(self.skills)

````python
class Employee:

    def __init__(self, name, skills):
        self.name = name
        self.skills = skills

    def __len__(self):
        return len(self.skills)

employee = Employee(
    "Manish",
    ["Python", "SQL", "Machine Learning"]
)

print(len(employee))
````

## 4️⃣ __eq__() — Object Comparison

## __eq__()

`__eq__()` defines how two objects should be compared using `==`.

Example:

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __eq__(self, other):
        return self.name == other.name and self.age == other.age

````python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __eq__(self, other):
        return self.name == other.name and self.age == other.age

student1 = Student("Manish", 21)
student2 = Student("Manish", 21)
student3 = Student("Rohan", 22)

print(student1 == student2)
print(student1 == student3)
````

## __iter__()

`__iter__()` makes a custom object iterable so that it can be used in a `for` loop.

Example:

```python
class Student:

    def __init__(self, subjects):
        self.subjects = subjects

    def __iter__(self):
        return iter(self.subjects)

````python
for item in ["Python", "SQL", "ML"]:
    print(item)
    
````

````python
class Student:

    def __init__(self, subjects):
        self.subjects = subjects

    def __iter__(self):
        return iter(self.subjects)

student = Student(["Python", "SQL", "Machine Learning"])

for subject in student:
    print(subject)
````

## __setitem__()

`__setitem__()` defines the behavior of assigning a value using `[]`.

Example:

```python
class Employee:

    def __init__(self, skills):
        self.skills = skills

    def __setitem__(self, index, value):
        self.skills[index] = value

````python
class Employee:

    def __init__(self, skills):
        self.skills = skills

    def __getitem__(self, index):
        return self.skills[index]

    def __setitem__(self, index, value):
        self.skills[index] = value

emp = Employee(["Python", "SQL", "Pandas"])

print(emp[0])

emp[0] = "Python Advanced"

print(emp[0])
````
