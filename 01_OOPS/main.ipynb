# Types of Methods in Python

Python classes commonly use three types of methods:

1. Instance Method
2. Class Method
3. Static Method

## 1. Instance Method

An **instance method** is a method that works with a specific object/instance.

It normally takes `self` as its first parameter.

### Example

```python
class Student:
    def __init__(self, name):
        self.name = name

    def study(self, subject):
        print(f"{self.name} is studying {subject}")

````python
class Student:

    def __init__(self, name):
        self.name = name

    def study(self, subject):
        print(f"{self.name} is studying {subject}")

student1 = Student("Manish")
student1.study("python")
````

## 2. Class Method

A **class method** is a method that works with the class rather than a specific instance.

It is created using the `@classmethod` decorator.

The first parameter is conventionally called `cls`.

### Example

```python
class Student:
    college = "KCC College"

    @classmethod
    def change_college(cls, new_college):
        cls.college = new_college

````python
class Student:
    college = "KCC College"

    @classmethod
    def show_college(cls):    ## clss==> mtlb class ke attribute ko use kr skta hai 
        print(cls.college)

Student.show_college()   # jab hum chahenge tab execute
````

## 3. Static Method

A **static method** is a method that does not automatically receive the instance or class as its first argument.

It is created using the `@staticmethod` decorator.

### Example

```python
class Student:

    @staticmethod
    def is_adult(age):
        return age >= 18

````python
class Student:

    @staticmethod
    def is_adult(age):
        return age >= 18
print(Student.is_adult(21))
````

## Name Mangling

Name Mangling is a mechanism in Python that changes the name of an attribute or method starting with two underscores (__) by adding the class name to its name. It is mainly used to prevent accidental access or overriding of private members.

````python
class Student:
    def __init__(self,name,age):
        self.__name = name          ## private
        self.__age= age                ## private
    def show_detials(self):
        print(f"Name: {self.__name}")
        print(f"Age: {self.__age}")

student1 = Student("manish",34)
student1.show_detials()
````

## Setter

A **setter** is a method used to modify or update the value of an attribute.

A setter can also validate the new value before updating the attribute.

```python
