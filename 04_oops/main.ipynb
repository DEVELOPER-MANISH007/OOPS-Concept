# Abstraction

Abstraction is the process of hiding unnecessary implementation details and exposing only the essential interface.

### Real-world example

A car provides:

- Start
- Brake
- Accelerate

The user does not need to know the internal engine implementation.

### Abstraction in Python

Python provides the `abc` module for implementing abstraction.

```python
from abc import ABC, abstractmethod

````python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self):
        pass
    
class UPI(Payment):
    def pay(self):
        print("Payment through UPI")

upi = UPI()
upi.pay()
````

````python
from abc import ABC, abstractmethod

class BankAPP(ABC):
    ## define abstractmethod
    @abstractmethod
    def security(self):
        pass
    ## define own method that you want
    def database(self):
        print("Connected to The Database")

class MobileApp(BankAPP):
    def login_mobile(self):
         print("Login into Mobile")
    def security(self):
        return print("security is Implemented")





mob = MobileApp()

mob.login_mobile()
````

# 🧪 PRACTICE 1 — Abstraction



````python
from abc import ABC,abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
    
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
        
    def area(self):
        return 3.14 * self.radius * self.radius

class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
         return self.length * self.width

        
        
    
circle = Circle(5)
rectangle = Rectangle(10, 5)

print(circle.area())
print(rectangle.area())
````

## Abstract Class with Normal Methods

An abstract class can contain:

- Abstract methods
- Normal methods
- Constructor (`__init__`)
- Attributes

Only abstract methods must be implemented by concrete child classes.

Example:

```python

class Model(ABC):

    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(f"Model: {self.name}")

    @abstractmethod
    def predict(self):
        pass

> Abstract class mein abstract methods ke saath normal methods aur __init__() bhi ho sakte hain.

````python
from abc import ABC, abstractmethod

class Model(ABC):

    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(f"Model: {self.name}")

    @abstractmethod
    def predict(self):
        pass

class LinearRegression(Model):

    def predict(self):
        print(f"{self.name} is predicting")



lr = LinearRegression("Linear Regression")

lr.show_name()
lr.predict()
````

## Multiple Abstract Methods

An abstract class can contain multiple abstract methods.

Every concrete child class must implement all abstract methods.

Example:

```python
from abc import ABC, abstractmethod

class Model(ABC):

    @abstractmethod
    def predict(self):
        pass

    @abstractmethod
    def evaluate(self):
        pass

````python
from abc import ABC, abstractmethod


class Model(ABC):

    @abstractmethod
    def predict(self):
        pass

    @abstractmethod
    def evaluate(self):
        pass

class LinearRegression(Model):

    def predict(self):
        print("Linear Regression is predicting")

    def evaluate(self):
        print("Evaluating Linear Regression")

lr = LinearRegression()
lr.predict()
lr.evaluate()
````

# 🧪 PRACTICE 4 — ML Model Contract 🔥

Ab ek realistic Data Science example.

### Abstract class:

```text
Model

````python
from abc import ABC, abstractmethod


class Model(ABC):

    @abstractmethod
    def predict(self):
        pass

    @abstractmethod
    def train(self):
        pass

    @abstractmethod
    def evaluate(self):
        pass


class LinearRegression(Model):

    def train(self):
        print("Training Linear Regression")

    def predict(self):
        print("Predicting with Linear Regression")

    def evaluate(self):
        print("Evaluating Linear Regression")


class RandomForest(Model):

    def train(self):
        print("Training Random Forest")

    def predict(self):
        print("Predicting with Random Forest")

    def evaluate(self):
        print("Evaluating Random Forest")


lr = LinearRegression()
rf = RandomForest()

lr.train()
lr.predict()
lr.evaluate()

rf.train()
rf.predict()
rf.evaluate()
````

````python
from abc import ABC, abstractmethod

class Model(ABC):

    @abstractmethod
    def predict(self):
        print("Generic model prediction")
        
class LinearRegression(Model):

    def predict(self):
        super().predict()
        print("Linear Regression prediction")

lr = LinearRegression()
lr.predict()
````
