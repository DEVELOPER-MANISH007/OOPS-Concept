## 1. Errors and Exceptions

An **exception** is an event that interrupts the normal execution of a Python program.

Common exceptions include:
- `ZeroDivisionError` — dividing by zero
- `ValueError` — invalid value
- `TypeError` — incompatible types
- `IndexError` — invalid list index
- `KeyError` — missing dictionary key
- `FileNotFoundError` — requested file does not exist

```python
# Example of an exception
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

## 2. The `try` and `except` Blocks

Put code that may cause an exception inside `try`. If an exception occurs, `except` handles it.

```python
try:
    number = int(input("Enter a number: "))
    print("You entered:", number)
except ValueError:
    print("Please enter a valid integer.")
```

## 3. Handling Specific Exceptions

It is usually better to catch the specific exception you expect instead of catching every possible exception.

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    print("Result:", a / b)
except ValueError:
    print("Only numbers are allowed.")
except ZeroDivisionError:
    print("The second number cannot be zero.")
```

## 4. Multiple Exceptions in One `except`

When multiple exceptions should be handled in the same way, they can be grouped in a tuple.

```python
try:
    value = int(input("Enter a number: "))
    print(100 / value)
except (ValueError, ZeroDivisionError):
    print("Invalid input or division by zero.")
```

## 5. Using `as` to Access the Exception

`except Exception as e` stores the exception object in `e`, allowing you to inspect its message.

```python
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("Error:", e)
```

## 6. The `else` Block

The `else` block runs **only when no exception occurs** in the `try` block.

```python
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid number.")
else:
    print("Successfully converted:", number)
```

## 7. The `finally` Block

The `finally` block runs whether an exception occurs or not. It is useful for cleanup operations.

```python
try:
    print("Opening resource...")
    result = 20 / 4
    print("Result:", result)
except ZeroDivisionError:
    print("Division error")
finally:
    print("Cleanup completed.")
```

## 8. Complete `try-except-else-finally` Example

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
except ValueError:
    print("Please enter an integer.")
except ZeroDivisionError:
    print("Zero is not allowed.")
else:
    print("Result:", result)
finally:
    print("Program finished.")
```

## 9. Raising Exceptions with `raise`

Use `raise` when your program needs to deliberately report an invalid condition.

```python
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")
    return age

try:
    print(check_age(-5))
except ValueError as e:
    print("Error:", e)
```

## 10. Custom Exceptions

You can create your own exception class by inheriting from `Exception`.

```python
class InsufficientBalanceError(Exception):
    pass


def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientBalanceError("Insufficient balance.")
    return balance - amount


try:
    balance = withdraw(1000, 1500)
    print("Remaining balance:", balance)
except InsufficientBalanceError as e:
    print("Transaction failed:", e)
```

## 11. Practical Example — Safe Calculator

```python
def calculator():
    try:
        a = float(input("Enter first number: "))
        operator = input("Enter operator (+, -, *, /): ")
        b = float(input("Enter second number: "))

        if operator == "+":
            result = a + b
        elif operator == "-":
            result = a - b
        elif operator == "*":
            result = a * b
        elif operator == "/":
            result = a / b
        else:
            raise ValueError("Invalid operator.")

        print("Result:", result)
    except ValueError as e:
        print("Input error:", e)
    except ZeroDivisionError:
        print("Cannot divide by zero.")

calculator()
```

## 12. Best Practices

- Catch specific exceptions whenever possible.
- Keep the `try` block small and focused.
- Use `finally` for cleanup tasks.
- Use meaningful custom exceptions for application-specific problems.
- Do not silently ignore exceptions.
- Avoid using a bare `except:` unless there is a specific reason.
- Give users clear and useful error messages.
